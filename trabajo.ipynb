```python
# Celda de verificación del entorno virtual
# Muestra el intérprete usado por el kernel, la versión de pip y comprueba imports claves
import sys, subprocess, importlib
print('sys.executable ->', sys.executable)
try:
    out = subprocess.check_output([sys.executable, '-m', 'pip', '--version'], text=True)
    print('pip ->', out.strip())
except Exception as e:
    print('No se pudo consultar pip:', e)

paquetes = ['pandas', 'unidecode', 'matplotlib', 'PIL']
for pkg in paquetes:
    try:
        m = importlib.import_module(pkg if pkg != 'PIL' else 'PIL')
        ver = getattr(m, '__version__', None)
        # algunas librerías reportan versión en otro atributo
        if ver is None:
            ver = getattr(m, 'VERSION', getattr(m, 'version', 'desconocida'))
        print(f"{pkg}: instalado, versión = {ver}")
    except Exception as e:
        print(f"{pkg}: NO instalado ({e})")
```

```python
# Intentar instalar paquetes faltantes en el intérprete del kernel
# Esto usa el mismo Python que ejecuta la libreta (sys.executable)
import sys, subprocess
packages = ['unidecode', 'matplotlib']
for pkg in packages:
    try:
        print(f'Instalando {pkg} en', sys.executable)
        subprocess.check_call([sys.executable, '-m', 'pip', 'install', pkg])
        print(f'{pkg} instalado correctamente')
    except Exception as e:
        print(f'Error instalando {pkg}:', e)

# Verificación rápida de imports
for pkg in packages:
    try:
        __import__(pkg)
        print(pkg + ' -> import OK')
    except Exception as e:
        print(pkg + ' -> import ERROR:', e)
```

```python
# Comprensión de los datos
import pandas as pd

df = pd.read_csv('./pictogramas_dataset/dataset_pictogramas.csv')
df.info()
```

```python
# Eliminar posibles duplicados
df_clean = df.drop_duplicates()
df_clean.info()
```

```python
# Longitud del dataset
print(f"Longitud del dataset limpio: {len(df_clean)}")
```

```python
# Número de pictogramas por categoría
category_counts = df_clean['categoria'].value_counts()
print(category_counts)
```

```python
# Visualizar cinco pictogramas de ejemplo por categoría
import matplotlib.pyplot as plt
import random
from PIL import Image
import os
categories = df_clean['categoria'].unique()
for category in categories:
    sample_images = df_clean[df_clean['categoria'] == category].sample(5) # Cinco por categoría
    plt.figure(figsize=(15, 3))
    for i, (_, row) in enumerate(sample_images.iterrows()):
        img_path = os.path.join('./pictogramas_dataset', row['archivo']) # Une 
        # la ruta base con el nombre del archivo
        img = Image.open(img_path)
        plt.subplot(1, 5, i + 1) # Muestra las imágenes en una fila de cinco columnas
        plt.imshow(img)
        plt.axis('off')
        plt.title(row['palabra']) # Agregamos el título con la palabra   
    plt.suptitle(f'Categoría: {category}')
    plt.show()
```

```python
# NORMALIZACIÓN DE TEXTO
# Convertir todas las palabras a minúsculas
from unidecode import unidecode
df_clean['palabra'] = df_clean['palabra'].str.lower().apply(unidecode) # Quitamos tildes por facilidad
```
